# function parameter

```python
def say_hello(user_name):
    print(f"hello {user_name}")

say_hello("alice")
say_hello("bob")
```

```python
def say_hello(user_name, *args, **kargs): # multiple arguments
```

```python
def say_hello(user_name, fav_drink):
    print(f"hello {user_name}, would like {fav_drink}")

say_hello("alice", "Tea")
say_hello("bob")
```
