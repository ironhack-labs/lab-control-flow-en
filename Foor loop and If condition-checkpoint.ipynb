```python
# Foor Loop 
names = ["Albandari", "Sara", "Laila", "Khalid", "Mona", "Ahmed", "Fatima", "Huda", "Samira", "Noura"]
names_ending_with_a = []
for name in names:
    if name.endswith('a'):
        names_ending_with_a.append(name)
        
print("Names ending with 'a':")
for name in names_ending_with_a:
    print(name)

```
