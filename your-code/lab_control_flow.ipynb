## Words

```python
words = ['play', 'filling', 'bar', 'theatre', 'easygoing', 'date', 'lead', 'that', 'story',  'island']
```

**Print every word in upper case**

```python
for i in words:
  print(i.upper())
```

**Create a new list containing only words with 5 or more letters**

```python
nl=[]

for i in words:
  if len(i)>=5:
    nl.append(i)

print(nl)
```

**Print the first word starting with "t"**

```python
for i in words:
  if i.startswith('t'):
    break

print(i)
```

## Numbers

**Create a list containing the square of every number from 1 to 10**

```python
list_sq = []

for i in range(1,11):
  list_sq.append(i**2)

print(list_sq)
```

**Print a list containing the square of every odd number from 1 to 10**

```python
odd_nm = []

for i in range(1,11):
  if i%2 != 0:
    odd_nm.append(i**2)

print(odd_nm)
```

**Create a list with the squares of all multiples of 8 below 1000**

```python
multi_eight = []

for i in range(1,1001):
  if i%8 == 0:
    multi_eight.append(i**2)

print(multi_eight)
```

## People

```python
people = [
    {
        "name": "Juan",
        "age": 34,
        "n_kids": 2
    },
    {
        "name": "Pepe",
        "age": 27,
        "n_kids": 0
    },
    {
        "name": "Sonia",
        "age": 41,
        "n_kids": 1
    },
    {
        "name": "Lucía",
        "age": 22,
        "n_kids": 2
    },
    {
        "name": "Leo",
        "age": 55,
        "n_kids": 5
    }
]
```

**How many people are there?**

```python
print(len(people))
```

**How many people have kids**?

```python
i = 0
for person in people:
  if person['n_kids'] > 0:
    i+=1

print(i)
```

**How many kids do they have in total?**

```python
total=0
for person in people:
  total+=person['n_kids']

print(total)
```

**In a year's time, names ending with "a" will have an extra kid. Create a list of dictionaries with people's info in a year's time**

```python
next_year = []

for person in people:
  updated_p = {
        'name': person['name'],
        'age': person['age']+1,
        'n_kids': person['n_kids']
    }
  if person['name'].endswith('a'):
    updated_p['n_kids']+=1
  next_year.append(updated_p)

print(next_year)
```
