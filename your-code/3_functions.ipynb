## Iteration 3:
Create a function called calculate_total() that:
Takes a list of numbers. Example: [5, 10, 15]
Returns the sum of all of them. Example: 30

```python
def calculate_total(numlist):
    total = 0
    for number in numlist:
        total = total + number
    
    return total

total = calculate_total([10,20,30])
# total = calculate_total(10,20,30) TypeError: calculate_total() takes 1 positional argument but 3 were given

print(f"total {total}")
```
